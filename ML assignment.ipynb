```python
df_clean["RatingGroup"] = (
    df_clean["EnergyRating"].str[0]
)

df_clean["RatingGroup"].value_counts()
```

```python
rating_map = {
    "A": "High",
    "B": "High",
    "C": "Medium",
    "D": "Medium",
    "E": "Low",
    "F": "Low",
    "G": "Low"
}

df_clean["EfficiencyClass"] = (
    df_clean["RatingGroup"].map(rating_map)
)

df_clean["EfficiencyClass"].value_counts()
```

```python
df_clean["CountyName"].value_counts().head(10)
```

```python
print("Unique county names:",
      df_clean["CountyName"].nunique())

print(df_clean["CountyName"].sort_values().unique())
```

```python
df_clean["CountyClean"] = (
    df_clean["CountyName"]
    .str.replace(r"^Co\.\s*", "", regex=True)
    .str.replace(r"\s+City$", "", regex=True)
    .str.replace(r"^Dublin\s+\d+$", "Dublin", regex=True)
    .str.strip()
)
```

```python
print("Before:", df_clean["CountyName"].nunique())
print("After:", df_clean["CountyClean"].nunique())

print(df_clean["CountyClean"].sort_values().unique())
```

```python
df_clean["CountyClean"] = (
    df_clean["CountyClean"]
    .str.replace(r"^Dublin\s+\d+[A-Za-z]$", "Dublin", regex=True)
)

print("Unique counties:", df_clean["CountyClean"].nunique())
```
